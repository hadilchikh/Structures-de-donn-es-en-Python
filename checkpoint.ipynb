```python
# Question 2
liste = [(2, 5), (1, 2), (4, 4), (2, 3), (2, 1)]

liste.sort(key=lambda x: x[-1])

print(liste)
```

```python
# Question 3
d1 = {'a': 100, 'b': 200, 'c': 300}

d2 = {'a': 300, 'b': 200, 'd': 400}

d3 = d1.copy()

for cle, valeur in d2.items():
    d3[cle] = d3.get(cle, 0) + valeur

print(d3)
```

```python
# Question 4
n = int(input("Entrez un nombre : "))

d = {}

for i in range(1, n + 1):
    d[i] = i * i

print(d)
```

```python
# Question 5
liste = [
    ('item1', '12.20'),
    ('item2', '15.10'),
    ('item3', '24.5')
]

liste.sort(key=lambda x: float(x[1]), reverse=True)

print(liste)
```

```python
# Question 6

# Créer un ensemble
ensemble = {0, 1, 2, 3, 4}

print(ensemble)

# Parcourir un ensemble
for element in ensemble:
    print(element)

# Ajouter un élément
ensemble.add(5)

print(ensemble)

# Supprimer un élément
ensemble.remove(2)

print(ensemble)
```
